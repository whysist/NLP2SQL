```python
import os
from langchain.agents import AgentState, create_agent
from langchain.tools import tool
from langchain_core.prompts import PromptTemplate,ChatPromptTemplate,MessagesPlaceholder
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
def add(a:int, b:int)->int:
    """Add two numbers."""
    return a+b
def mul(a:float,b:float)->float:
    """Mutliply two numbers."""
    return a*b
def power(a:float,b:float)->float:
    """Find a to the power of b."""
    return a**b

tool=[add,mul,power]
```

```python
prompt=ChatPromptTemplate.from_messages(
    [
    ("system",
        """
        You are a mathematical assistant.
        Use the available tools to answer the question, If no suitable 
        tool exists, inform promptly,
        
        Return ONLY the FINAL ANSWER.
        
        Example:
        Human: what is 4+7?
        AI: 11
        """
    ),
    MessagesPlaceholder("chat_history",optional=True),
    ("human","{input}"),
    MessagesPlaceholder("agent_scratchpad"),
    
    ]
)

custom_instructions = """
You are a mathematical assistant.
Use the available tools to answer the question. If no suitable 
tool exists, inform the user promptly.

Return ONLY the final answer.
"""
```

```python
Llm=ChatGoogleGenerativeAI(model="gemini-2.5-flash",temperature=0)
agent=create_agent(Llm,tools=tool,system_prompt=custom_instructions)
```

```python
result = agent.invoke({
    "messages": [{"role": "user", "content": "What is 2 to the power of 1 ?"}]
})
print(result["messages"][-1].content)
```
