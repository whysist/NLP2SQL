```python
import os
from dotenv import load_dotenv
from langchain.tools import tool
```

```python
@tool
def tool1(x:float)->float:
    """Calculate the square root of a number"""
    return x**0.5

```

```python
tool1.invoke({"x":467})   
#this is a local function not an agentic one
```

### Using it in an Agent

```python
load_dotenv()
```

```python
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model


model=init_chat_model(model="gemini-2.5-flash",model_provider="google_genai")

agent=create_agent(model=model,
                   tools=[tool1],
                   system_prompt="""
                   You are an arithmetic wizard. use your tools to calculate the 
                   square root of a number""")
```

```python
from langchain.messages import HumanMessage,AIMessage
from pprint import pprint
```

```python
question=HumanMessage("What's the square root of 467 ?")

response=agent.invoke(
    {'messages':[question]}
)
```

```python
pprint(response)
print(response['messages'][-1].content)
print(response['messages'][1].tool_calls)
```

```python
from typing import Dict,Any
from tavily import TavilyClient
from langchain.tools import tool
```

```python
load_dotenv()
```

```python
tavily_client=TavilyClient()

@tool
def web_search(query:str)->Dict[str,Any]:
    """Search the web for info"""
    return tavily_client.search(query)


# web_search.invoke("Who's the current mayor of San Francisco ?")
```

```python
model1=init_chat_model(model="gemini-2.5-flash-lite",model_provider="google_genai")

agent1=create_agent(model=model1,tools=[web_search])

question=HumanMessage(content="Who is the current mayor of San Franciso ?")

response=agent1.invoke(
    {"messages":[question]}
)

print(response['messages'][-1].content)
```
