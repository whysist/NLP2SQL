```python
import os
from langchain.agents import create_agent,AgentState
from langchain_core.prompts import ChatPromptTemplate,MessagesPlaceholder
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import SystemMessage
from langchain_core.tools import tool
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
# @tool
def add(a:int, b:int)->int:
    """Add two numbers."""
    return a+b

# @tool
def mul(a:float,b:float)->float:
    """Mutliply two numbers."""
    return a*b

# @tool
def power(a:float,b:float)->float:
    """Find a to the power of b."""
    return a**b

tools=[add,mul,power]
```

```python
# p1=ChatPromptTemplate.from_messages([
#     ("system",
#      """You are a mathematical assistant. Solve the user's problems using the available 
#      tools and return ONLY the final answer
#      """),
#     ("human","{input}")
# ])


prompt=SystemMessage(content="""You are a mathematical assistant.
                     Solve the user's problem using the available tools and 
                     return error string if not available. 
                     Return ONLY the final answer
                     """)
```

```python
from dataclasses import dataclass

@dataclass
class Response:
    output:int  #The LLM can change the required type based on the response
```

```python
Llm=ChatGoogleGenerativeAI(model="gemini-2.5-flash",temperature=0)
agent=create_agent(Llm,tools=tools,system_prompt=prompt,response_format=Response)
```

```python
try:
    result=agent.invoke({
        "messages":{"role":"user","content":"What is the 15-9 ?"}
    })
except Exception as e:
    print(e)

print(result["structured_response"].output)
print(type(result["structured_response"].output))
```
